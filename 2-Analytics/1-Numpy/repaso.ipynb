```python
def mi_funcion(a,b,c,d = 10, *args, **kwargs):
    print("a -->", a)
    print("b -->", b)
    print("c -->", c)
    print("d -->", d)
    print("args -->", args)
    print("kwargs -->", kwargs)
    
def otra_funcion(a,b,/,c,d, *, e=10, f=20):
    print("a", a)
    print("b", b)
    print("c", c)
    print("d", d)
    print("e", e)
    print("f", f)
    
otra_funcion(1,2,c=3,d=4, f=100)
```
